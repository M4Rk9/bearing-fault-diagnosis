# STFT + DWT + Envelope: Hybrid CNN–LSTM
Train and compare both models on identical recording-isolated partitions. Run from the repository root or change the working directory first. Do not interpret results until the real-data checks and training succeed.

```python
from pathlib import Path
import os, sys, subprocess
if not Path('src/train_hybrid.py').exists() and Path('../src/train_hybrid.py').exists():
    os.chdir('..')
assert Path('src/train_hybrid.py').exists(), 'Open the notebook in the repository'
# Change to the location of your numeric ZIP archive (not the images ZIP).
DATASET = Path(r'Bearing_Fault_Feature_Dataset(1).zip').resolve()
OUTPUT = Path('models/hybrid_experiment').resolve()
assert DATASET.exists(), f'Missing: {DATASET}'
EPOCHS = 30
```

```python
# Validate the real data before training.
subprocess.run([sys.executable, '-m', 'src.run_hybrid_experiment', '--dataset', str(DATASET), '--check-only'], check=True)
```

```python
# Train both models, save trained weights, evaluate, and generate the report.
subprocess.run([sys.executable, '-m', 'src.run_hybrid_experiment', '--dataset', str(DATASET), '--epochs', str(EPOCHS), '--output', str(OUTPUT)], check=True)
```

```python
import pandas as pd
from IPython.display import display, Image, Markdown
report = OUTPUT / 'report'
display(pd.read_csv(report / 'metrics.csv'))
display(pd.read_csv(report / 'per_class_metrics.csv'))
for figure in sorted(report.glob('*.png')):
    display(Image(filename=str(figure)))
display(Markdown((report / 'results_and_discussion.md').read_text(encoding='utf-8')))
```
